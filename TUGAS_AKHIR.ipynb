# Tugas Akhir: Prediksi Harga Saham dengan LSTM
Menggunakan pengetahuan tentang RNN dan LSTM, kita akan membangun model deep learning untuk memprediksi harga saham berdasarkan dataset time series dari Kaggle: [Stock Time Series Dataset](https://www.kaggle.com/datasets/szrlee/stock-time-series-20050101-to-20171231)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense

df = pd.read_csv("all_stocks_2006-01-01_to_2018-01-01.csv")
df["Date"] = pd.to_datetime(df["Date"])
df.set_index("Date", inplace=True)

def create_sequences(data, window_size=30):
    sequences, targets = [], []
    for i in range(len(data) - window_size):
        sequences.append(data[i:i+window_size])
        targets.append(data[i+window_size])
    return np.array(sequences), np.array(targets)

names = df["Name"].unique()
results = {}

for name in names:
    print(f"\n🟢 Processing {name}...")
    company_df = df[df["Name"] == name][["Close"]].copy()
    
    if len(company_df) < 100:
        print(f"🔸 Skip {name}, data terlalu sedikit.")
        continue
    
    # Split data sebelum scaling
    split_idx = int(len(company_df) * 0.8)
    train_data = company_df.iloc[:split_idx]
    test_data = company_df.iloc[split_idx:]
    
    # Scaling dengan data training
    scaler = MinMaxScaler()
    train_scaled = scaler.fit_transform(train_data)
    test_scaled = scaler.transform(test_data)
    
    # Gabungkan data scaled untuk membuat sequences
    all_scaled = np.concatenate([train_scaled, test_scaled]).flatten()
    
    window_size = 30
    X, y = create_sequences(all_scaled, window_size)
    X = X[..., np.newaxis]  # Menambahkan dimensi fitur
    
    # Split sequences
    train_seq_end = split_idx - window_size
    X_train, X_val = X[:train_seq_end], X[train_seq_end:]
    y_train, y_val = y[:train_seq_end], y[train_seq_end:]
    
    # Model LSTM
    model = Sequential([
        LSTM(64, return_sequences=False, input_shape=(window_size, 1)),
        Dense(1)
    ])
    
    model.compile(loss='mse', optimizer='adam')
    history = model.fit(X_train, y_train, 
                      validation_data=(X_val, y_val),
                      epochs=50, 
                      verbose=0,
                      batch_size=32)
    
    val_loss = model.evaluate(X_val, y_val, verbose=0)
    results[name] = val_loss
    print(f"✅ {name} validation loss: {val_loss:.6f}")
    
    # Prediksi
    y_pred = model.predict(X_val)
    
    # Plotting
    plt.figure(figsize=(10,4))
    plt.plot(y_val, label="Actual")
    plt.plot(y_pred, label="Predicted")
    plt.title(f"{name} - Validation Prediction")
    plt.legend()
    plt.show()
    
    # Forecasting
    n_forecast = 7
    last_sequence = all_scaled[-window_size:]
    input_seq = last_sequence.reshape(1, window_size, 1)
    forecast = []
    
    for _ in range(n_forecast):
        next_pred = model.predict(input_seq, verbose=0)[0, 0]
        forecast.append(next_pred)
        input_seq = np.append(input_seq[:, 1:, :], [[[next_pred]]], axis=1)
    
    forecast_prices = scaler.inverse_transform(np.array(forecast).reshape(-1, 1)).flatten()
    
    # Generate dates
    last_date = company_df.index[-1]
    forecast_dates = pd.date_range(
        start=last_date + pd.Timedelta(days=1),
        periods=n_forecast
    )
    
    # Plot forecast
    plt.figure(figsize=(10,5))
    plt.plot(company_df.index[-100:], company_df["Close"].values[-100:], label="Historis")
    plt.plot(forecast_dates, forecast_prices, 'ro-', label="Prediksi 7 Hari")
    plt.title(f"Prediksi Harga Saham {name} - 7 Hari ke Depan")
    plt.xlabel("Tanggal")
    plt.ylabel("Harga")
    plt.legend()
    plt.tight_layout()
    plt.show()

# Hasil akhir
print("\n📊 Hasil Validasi:")
for k, v in sorted(results.items(), key=lambda x: x[1]):
    print(f"{k}: {v:.6f}")
```

```python
sorted_results = sorted(results.items(), key=lambda x: x[1])
names = [k for k, _ in sorted_results]
losses = [v for _, v in sorted_results]

plt.figure(figsize=(12, 6))
plt.bar(names, losses)
plt.xticks(rotation=45)
plt.ylabel("Validation Loss (MSE)")
plt.title("Validation Loss per Company")
plt.tight_layout()
plt.show()
```
